 Exercício 3 — Depuração

```python
class Node:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None


n1 = Node(10)
n2 = Node(20)
n3 = Node(30)

n1.proximo = n2
n2.proximo = n3

atual = n1

while atual is not None:
    print(atual.valor)
    atual = atual.proximo
```

**Qual é o problema?**
O problema é que falta atualizar a variável atual dentro do while. Assim, ela continua apontando para o mesmo nó.

**Por que o programa não termina?**
Porque atual continua sendo n1, então atual nunca chega a None. O while fica repetindo para sempre.

**Qual linha deve ser acrescentada?**
Dentro do while, deve ser acrescentada: atual = atual.proximo

**Qual será a saída depois da correção?**
10
20
30

Problema real — Sistema de atendimento de uma clínica

```python
class Paciente:
    def __init__(self, nome, idade, prioridade):
        self.nome = nome
        self.idade = idade
        self.prioridade = prioridade


class Node:
    def __init__(self, paciente):
        self.paciente = paciente
        self.proximo = None


class FilaAtendimento:
    def __init__(self):
        self.inicio = None
        self.fim = None
        self.quantidade = 0

    # 1. Adicionar paciente
    def adicionar(self, paciente):
        novo = Node(paciente)

        if self.fim is None:
            self.inicio = novo
            self.fim = novo
        else:
            self.fim.proximo = novo
            self.fim = novo

        self.quantidade += 1

    # 2. Listar pacientes
    def listar(self):
        atual = self.inicio

        while atual is not None:
            paciente = atual.paciente
            print(paciente.nome, "-", paciente.idade, "anos -", paciente.prioridade)
            atual = atual.proximo

    # 3. Atender o primeiro paciente
    def atender(self):
        if self.inicio is None:
            return None

        paciente = self.inicio.paciente
        self.inicio = self.inicio.proximo

        if self.inicio is None:
            self.fim = None

        self.quantidade -= 1

        return paciente

    # 4. Verificar se está vazia
    def vazia(self):
        return self.inicio is None

    # 5. Informar quantidade
    def tamanho(self):
        return self.quantidade


fila = FilaAtendimento()

fila.adicionar(Paciente("Ana", 32, "Normal"))
fila.adicionar(Paciente("Bruno", 70, "Prioridade"))
fila.adicionar(Paciente("Carlos", 45, "Normal"))

fila.listar()

paciente = fila.atender()

print("Atendendo:", paciente.nome)
print("Quantidade restante:", fila.tamanho())
```
