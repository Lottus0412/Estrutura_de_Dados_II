Excercio 1 - O simulador "Desfazer"

```python
# Pilha (lista) que guarda os números adicionados
pilha = []

def adicionar(numero):
    pilha.append(numero)  # empilha (push)
    print(f"Adicionou {numero} -> Pilha atual: {pilha}")

def desfazer():
    if not pilha:
        print("Nada para desfazer.")
        return
    numero_removido = pilha.pop()  # desempilha (pop) - remove o último
    print(f"Desfez a adição de {numero_removido} -> Pilha atual: {pilha}")


#Testando
adicionar(5)
adicionar(10)
adicionar(3)

print("\nDesfazendo")
desfazer()  # remove o 3
desfazer()  # remove o 10
desfazer()  # remove o 5
desfazer()  # pilha vazia
```

Excercio 2 - O sistema de Impressão

```python
# Fila (lista) que guarda os documentos aguardando impressão
fila = []

def novo_documento(numero):
    fila.append(numero)  # entra no FINAL da fila
    print(f"Documento {numero} enfileirado -> Fila atual: {fila}")

def imprimir():
    if not fila:
        print("Nenhum documento na fila.")
        return
    documento = fila.pop(0)  # remove do INÍCIO da fila (o mais antigo)
    print(f"Imprimindo documento {documento} -> Fila atual: {fila}")


# Testando
novo_documento(1)
novo_documento(2)
novo_documento(3)

print("\nImprimindo")
imprimir()  # imprime o 1 (o mais antigo)
imprimir()  # imprime o 2
imprimir()  # imprime o 3
imprimir()  # fila vazia
```

Desafio Master

```python
from collections import deque

fila_atendimento = deque()

def paciente_normal(numero):
    fila_atendimento.append(numero)  # entra no FINAL da fila
    print(f"Paciente {numero} (normal) entrou na fila -> {list(fila_atendimento)}")

def paciente_prioritario(numero):
    fila_atendimento.appendleft(numero)  # fura a fila, entra no INÍCIO
    print(f"Paciente {numero} (PRIORIDADE) furou a fila -> {list(fila_atendimento)}")

def atender():
    if not fila_atendimento:
        print("Nenhum paciente na fila.")
        return
    paciente = fila_atendimento.popleft()  # sempre atende quem está na FRENTE
    print(f"Atendendo paciente {paciente} -> Fila atual: {list(fila_atendimento)}")


#Testando
paciente_normal(1)
paciente_normal(2)
paciente_normal(3)

print("\nChega uma emergência!")
paciente_prioritario(99)  # fura a fila

print("\nAtendimentos")
atender()  # atende o 99 primeiro (prioridade)
atender()  # atende o 1 (o normal mais antigo)
atender()  # atende o 2
atender()  # atende o 3
```
