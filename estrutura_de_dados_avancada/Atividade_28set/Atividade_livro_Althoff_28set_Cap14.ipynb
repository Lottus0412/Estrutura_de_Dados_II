```python
class BinaryTree:
  #Cria a classe que representa uma árvore binária
    def __init__(self, value):
        self.key = value
   #Armazena o valor do nó
        self.left_child = None
  #Defini inicialmente que o nó não possui filho a esquerda
        self.right_child = None
  #Defini inicialmente que o nó não possui filho a direita
    def insert_left(self, value):
        if self.left_child == None:
  #Verifica se o nó atual não possui um filho à esquerda
            self.left_child = BinaryTree(value)
  #Se não existe filho à esquerda, cria um novo nó
        else:
            bin_tree = BinaryTree(value)
   #Se existe filho à esquerda, cria um novo nó
            bin_tree.left_child = self.left_child
  #O filho esquerdo que já existia passa a ficar abaixo de um novo nó
            self.left_child = bin_tree
  #O novo nó passa a ser o filho esquerdo do nó atual

    def insert_right(self, value):
        if self.right_child == None:
  # Verifica se o nó atual não possui um filho à direita
            self.right_child = BinaryTree(value)
  # Se não existir, cria um novo nó com o valor informado
        else:
            bin_tree = BinaryTree(value)
  # Se já existir um filho à direita, cria um novo nó
            bin_tree.right_child = self.right_child
  # O filho direito que já existia passa a ficar abaixo do novo nó
            self.right_child = bin_tree
  # O novo nó passa a ser o filho direito do nó atual

tree = BinaryTree(1)
  # Cria a árvore e define o valor 1 como o nó raiz.

tree.insert_left(2)
  # Insere o valor 2 no lado esquerdo da raiz.

tree.insert_right(3)
  # Insere o valor 3 no lado direito da raiz.

tree.insert_left(4)
  # Insere o valor 4 no lado esquerdo.
  # Como já existe o nó 2, o nó 4 fica acima dele.

tree.left_child.insert_right(6)
  # Acessa o filho esquerdo da raiz (4) e insere o valor 6 à sua direita.

tree.insert_right(5)
  # Insere o valor 5 no lado direito.
  # Como já existe o nó 3, o nó 5 fica acima dele.
```

```python
#Atividade 1 (Ajuda da ia)
class BinaryTree:
    def __init__(self, value):
        self.key = value
        self.left_child = None
        self.right_child = None

    def insert_left(self, value):
        if self.left_child == None:
            self.left_child = BinaryTree(value)
        else:
            bin_tree = BinaryTree(value)
            bin_tree.left_child = self.left_child
            self.left_child = bin_tree

    def insert_right(self, value):
        if self.right_child == None:
            self.right_child = BinaryTree(value)
        else:
            bin_tree = BinaryTree(value)
            bin_tree.right_child = self.right_child
            self.right_child = bin_tree

    def has_leaf_nodes(self):
        if self.left_child == None and self.right_child == None:
            return True
        else:
            return False


tree = BinaryTree(1)

tree.insert_left(2)
tree.insert_right(3)
tree.insert_left(4)
tree.left_child.insert_right(6)
tree.insert_right(5)

print(tree.has_leaf_nodes())
```

```python
#Atividade 2 (Ajuda da Ia)

class BinaryTree:
    def __init__(self, value):
        self.key = value
        self.left_child = None
        self.right_child = None

    def insert_left(self, value):
        if self.left_child == None:
            self.left_child = BinaryTree(value)
        else:
            bin_tree = BinaryTree(value)
            bin_tree.left_child = self.left_child
            self.left_child = bin_tree

    def insert_right(self, value):
        if self.right_child == None:
            self.right_child = BinaryTree(value)
        else:
            bin_tree = BinaryTree(value)
            bin_tree.right_child = self.right_child
            self.right_child = bin_tree


def invert_tree(node):
    if node is None:
        return None
    node.left_child, node.right_child = node.right_child, node.left_child
    invert_tree(node.left_child)
    invert_tree(node.right_child)
    return node


tree = BinaryTree(1)
tree.insert_left(2)
tree.insert_right(3)
tree.insert_left(4)
tree.left_child.insert_right(6)
tree.insert_right(5)

invert_tree(tree)
```
